```python
from dotenv import load_dotenv
load_dotenv()
from google import genai

client = genai.Client()
MODEL = "gemini-3.8-flash"

interaction = client.interactions.create(
    model=MODEL,
    input="Explain how AI works in a few words",
    generation_config={"thinking_level": "low"},
)
print(interaction.output_text)

def generate_recipe(ingradients, cusine, diet):
    '''
    Generate a reciepe using ingradients provided.
    cusine can be "Italian", "Mexican", "Indian"
    diet can be "Vegen", "Jain", "Any"
    '''
    prompt = f'''
    Generate one fod reciepe using these ingradients : {','.join(ingradients)}
    Reciepe should not be more than 90 words
    Cusine:{cusine}
    Diet : {diet}
    '''
    response = client.interactions.create(
        model=MODEL,
        input=prompt,
        generation_config={"thinking_level": "low"},
    )
    return response.output_text

recipe = generate_recipe(["Tomato", "Rice", "chicken", "onions", "oil"], "Indian", "Any")
print(recipe)
```
